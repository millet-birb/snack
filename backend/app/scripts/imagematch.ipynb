```python
import pandas as pd
import re
from pathlib import Path

# =========================
# 파일 경로
# =========================
PRODUCT_FINAL_CSV = r"C:\Users\yooju\.vscode\safesnack\snack-safe-app\backend\app\data\최종데이터 전처리_final.csv"
IMAGE_CSV = r"C:\Users\yooju\.vscode\safesnack\snack-safe-app\backend\app\data\product_images.csv"

OUTPUT_DIR = Path("image_final_clean_output")
OUTPUT_DIR.mkdir(exist_ok=True)

product_final = pd.read_csv(PRODUCT_FINAL_CSV)
image_df = pd.read_csv(IMAGE_CSV)

# =========================
# 컬럼명
# =========================
PRODUCT_NAME_COL = "품목명"
PRODUCT_COMPANY_COL = "제조사명"

IMAGE_NAME_COL = "품목명"
IMAGE_COMPANY_COL = "제조사명"

print("product_final 행 수:", len(product_final))
print("product_images 원본 행 수:", len(image_df))

print("\n[product_final 컬럼]")
print(product_final.columns.tolist())

print("\n[product_images 컬럼]")
print(image_df.columns.tolist())
```

```python
# =========================
# 공통 함수
# =========================

def clean_space(text):
    return re.sub(r"\s+", " ", str(text)).strip()


def move_mini_front_in_wafer(name):
    """
    웨하스 + 미니가 같이 들어간 품목만 처리.
    '미니'를 웨하스 품목 내부에서 앞으로 이동.

    예:
    웨하스 미니 바닐라 -> 미니 웨하스 바닐라
    딸기 웨하스 미니 -> 미니 딸기 웨하스
    미니바닐라웨하스 -> 미니 바닐라 웨하스
    """
    if pd.isna(name):
        return name

    s = clean_space(name)

    if ("웨하스" not in s) or ("미니" not in s):
        return s

    # 붙어 있는 표현 분리
    s = re.sub(r"미니(?=[가-힣A-Za-z0-9])", "미니 ", s)
    s = re.sub(r"(?<=[가-힣A-Za-z0-9])미니", " 미니", s)

    s = re.sub(r"웨하스(?=[가-힣A-Za-z0-9])", "웨하스 ", s)
    s = re.sub(r"(?<=[가-힣A-Za-z0-9])웨하스", " 웨하스", s)

    s = clean_space(s)

    tokens = s.split()
    tokens_without_mini = [t for t in tokens if t != "미니"]

    new_name = "미니 " + " ".join(tokens_without_mini)

    return clean_space(new_name)


def preprocess_image_name_like_final(name, company=None):
    """
    product_final을 만들 때 적용했던 주요 전처리를
    product_images에도 적용하기 위한 함수.

    반환값:
    - 변경 후 품목명
    - 삭제 여부
    - 처리 사유
    """
    if pd.isna(name):
        return name, False, ""

    s = clean_space(name)
    company_text = "" if pd.isna(company) else str(company).strip()

    delete_flag = False
    reasons = []

    # ==================================================
    # 1. 기본 표기 통일
    # ==================================================
    replace_map = {
        "콘칲": "콘칩",
        "초코칲": "초코칩",
        "사브래": "사브레",
    }

    for old, new in replace_map.items():
        if old in s:
            s = s.replace(old, new)
            reasons.append(f"{old}->{new}")

    # ==================================================
    # 2. 카스타드 처리
    # ==================================================
    if s == "롯데제과 카스타드 오리지날 230g":
        delete_flag = True
        reasons.append("삭제: 롯데제과 카스타드 오리지날 230g")

    if "롯데" in company_text and "카스타드" in s and not s.startswith("롯데"):
        s = "롯데 " + s
        reasons.append("롯데 카스타드 prefix 추가")

    # ==================================================
    # 3. 고래밥 처리
    # ==================================================
    if ("왕고래밥" in s and "짜장라면" in s) or ("고래밥" in s and "짜장라면" in s):
        s = "짜장라면맛 고래밥"
        reasons.append("짜장라면맛 고래밥 표준화")

    if s == "왕고래밥 볶음양념맛":
        delete_flag = True
        reasons.append("삭제: 왕고래밥 볶음양념맛")

    # ==================================================
    # 4. 웨하스 + 미니 처리
    # 미니를 front로 이동
    # ==================================================
    before_wafer = s
    s = move_mini_front_in_wafer(s)

    if s != before_wafer:
        reasons.append("웨하스 미니 front 이동")

    # ==================================================
    # 5. 웨하스 제조사 prefix 처리
    # 중요:
    # 미니 이동 후 제조사 prefix를 붙이므로
    # 결과가 '서주 미니 웨하스 ...' 형태가 됨
    # ==================================================
    if "웨하스" in s:
        for maker in ["삼아", "서주", "해태"]:
            if maker in company_text and maker not in s:
                s = maker + " " + s
                reasons.append(f"{maker} 웨하스 prefix 추가")
                break

    # ==================================================
    # 6. 나쵸칩 처리
    # product_final 기준 최종명:
    # 피코크 나쵸칩 오리지널
    # ==================================================
    if s in ["나쵸칩 오리지날", "나쵸칩 오리지널", "나초칩 오리지날", "나초칩 오리지널"]:
        s = "피코크 나쵸칩 오리지널"
        reasons.append("피코크 나쵸칩 오리지널 표준화")

    elif s in ["나쵸칩", "나초칩"]:
        delete_flag = True
        reasons.append(f"삭제: {s}")

    # ==================================================
    # 7. 다이제샌드 처리
    # ==================================================
    if s == "다이제샌드":
        s = "다이제 샌드 바닐라&밀크"
        reasons.append("다이제샌드 -> 다이제 샌드 바닐라&밀크")

    # ==================================================
    # 8. 삭제 품목 처리
    # ==================================================
    if s in ["피자맛팝콘", "사브레"]:
        delete_flag = True
        reasons.append(f"삭제: {s}")

    # ==================================================
    # 9. 왕소라형과자 제조사 prefix
    # ==================================================
    if s == "왕소라형과자":
        if "청우" in company_text:
            s = "청우 왕소라형과자"
            reasons.append("청우 왕소라형과자 표준화")
        elif "코스모스" in company_text:
            s = "코스모스 왕소라형과자"
            reasons.append("코스모스 왕소라형과자 표준화")

    # ==================================================
    # 10. 해태 고구마형과자 삭제
    # ==================================================
    if s == "고구마형과자" and "해태" in company_text:
        delete_flag = True
        reasons.append("삭제: 해태 고구마형과자")

    s = clean_space(s)

    return s, delete_flag, " / ".join(reasons)
```

```python
# =========================
# 매칭용 정규화 함수
# =========================

def norm_name(text):
    if pd.isna(text):
        return ""

    s = str(text).lower().strip()

    replace_map = {
        "오리지널": "오리지날",
        "나초": "나쵸",
        "콘칲": "콘칩",
        "초코칲": "초코칩",
        "왕고래밥": "고래밥",
    }

    for old, new in replace_map.items():
        s = s.replace(old, new)

    s = re.sub(r"\([^)]*\)", " ", s)
    s = re.sub(r"\[[^\]]*\]", " ", s)
    s = re.sub(r"[^0-9a-z가-힣&]", "", s)

    return s


def norm_company(text):
    if pd.isna(text):
        return ""

    s = str(text).lower().strip()

    remove_map = {
        "주식회사": "",
        "(주)": "",
        "㈜": "",
        " ": "",
        ".": "",
        ",": "",
        "-": "",
        "_": "",
    }

    for old, new in remove_map.items():
        s = s.replace(old, new)

    company_replace_map = {
        "롯데웰푸드": "롯데",
        "롯데제과": "롯데",
        "해태제과식품": "해태",
        "해태제과": "해태",
        "크라운제과": "크라운",
        "오리온제과": "오리온",
        "청우식품": "청우",
        "코스모스제과": "코스모스",
        "삼아인터내셔날": "삼아",
        "서주제과": "서주",
        "서주푸드": "서주",
    }

    for old, new in company_replace_map.items():
        s = s.replace(old, new)

    s = re.sub(r"[^0-9a-z가-힣]", "", s)

    return s


def make_pair_key(name, company):
    """
    품목명 + 제조사명 기준 key
    """
    return norm_name(name) + "||" + norm_company(company)
```

```python
# =========================
# product_images에 final 기준 전처리 적용
# =========================

image_work = image_df.copy()

image_work["_image_original_row_id"] = image_work.index
image_work["_before_품목명"] = image_work[IMAGE_NAME_COL].astype(str).str.strip()

processed = image_work.apply(
    lambda row: preprocess_image_name_like_final(
        row[IMAGE_NAME_COL],
        row[IMAGE_COMPANY_COL]
    ),
    axis=1
)

image_work["_processed_품목명"] = [x[0] for x in processed]
image_work["_delete_flag"] = [x[1] for x in processed]
image_work["_preprocess_reason"] = [x[2] for x in processed]

# 품목명 변경 반영
image_work[IMAGE_NAME_COL] = image_work["_processed_품목명"]

# 삭제 flag가 True인 행 제거
image_preprocessed = image_work[
    ~image_work["_delete_flag"]
].copy()

print("product_images 원본 행 수:", len(image_df))
print("전처리 후 image_preprocessed 행 수:", len(image_preprocessed))
print("품목명 변경 행 수:", (image_work["_before_품목명"] != image_work["_processed_품목명"]).sum())
print("삭제 flag 행 수:", image_work["_delete_flag"].sum())

print("\n[품목명 변경 대상]")
display(
    image_work[
        image_work["_before_품목명"] != image_work["_processed_품목명"]
    ][
        ["_image_original_row_id", "_before_품목명", "_processed_품목명", "제조사명", "_preprocess_reason"]
    ].head(200)
)

print("\n[삭제 flag 대상]")
display(
    image_work[
        image_work["_delete_flag"]
    ][
        ["_image_original_row_id", "_before_품목명", "_processed_품목명", "제조사명", "_preprocess_reason"]
    ]
)
```

```python
# =========================
# product_final 기준으로 image 정렬/필터링
# =========================

product_base = product_final.copy()
image_base = image_preprocessed.copy()

product_base["_product_row_id"] = product_base.index
image_base["_image_row_id"] = image_base.index

# 품목명 + 제조사명 pair key
product_base["_pair_key"] = product_base.apply(
    lambda row: make_pair_key(row[PRODUCT_NAME_COL], row[PRODUCT_COMPANY_COL]),
    axis=1
)

image_base["_pair_key"] = image_base.apply(
    lambda row: make_pair_key(row[IMAGE_NAME_COL], row[IMAGE_COMPANY_COL]),
    axis=1
)

# 같은 pair_key가 여러 개 있을 수 있으므로 순번 부여
product_base["_pair_seq"] = product_base.groupby("_pair_key").cumcount()

image_base = image_base.sort_values("_image_original_row_id").copy()
image_base["_pair_seq"] = image_base.groupby("_pair_key").cumcount()

# image 쪽에서 붙일 컬럼
image_original_cols = image_df.columns.tolist()

image_attach = image_base[
    ["_pair_key", "_pair_seq", "_image_original_row_id", "_before_품목명", "_preprocess_reason"]
    + image_original_cols
].copy()

# image 컬럼명 충돌 방지
rename_map = {}
for col in image_original_cols:
    rename_map[col] = f"image_{col}"

image_attach = image_attach.rename(columns=rename_map)

# product_final 기준으로 merge
aligned = product_base.merge(
    image_attach,
    on=["_pair_key", "_pair_seq"],
    how="left"
)

aligned["이미지매칭여부"] = aligned["_image_original_row_id"].notna().map({
    True: "Y",
    False: "N"
})

print("product_final 행 수:", len(product_final))
print("aligned 행 수:", len(aligned))
print("행 수 일치 여부:", len(product_final) == len(aligned))

print("이미지 매칭 성공 수:", (aligned["이미지매칭여부"] == "Y").sum())
print("이미지 매칭 실패 수:", (aligned["이미지매칭여부"] == "N").sum())

display(
    aligned[
        [
            PRODUCT_NAME_COL,
            PRODUCT_COMPANY_COL,
            "image_품목명",
            "image_제조사명",
            "_before_품목명",
            "_preprocess_reason",
            "이미지매칭여부"
        ]
    ].head(5)
)
```

```python
# =========================
# 최종 image CSV 생성
# =========================

image_final = pd.DataFrame()

# 품목명/제조사명은 product_final 기준으로 강제 통일
image_final[IMAGE_NAME_COL] = product_final[PRODUCT_NAME_COL]
image_final[IMAGE_COMPANY_COL] = product_final[PRODUCT_COMPANY_COL]

# 나머지 image 컬럼은 매칭된 image에서 가져오기
for col in image_df.columns:
    if col in [IMAGE_NAME_COL, IMAGE_COMPANY_COL]:
        continue

    image_final[col] = aligned[f"image_{col}"]

print("product_final 행 수:", len(product_final))
print("image_final 행 수:", len(image_final))
print("행 수 일치 여부:", len(product_final) == len(image_final))

print("\nimage_url 없는 행 수:")
if "image_url" in image_final.columns:
    print(image_final["image_url"].isna().sum())
else:
    print("image_url 컬럼 없음")

display(image_final.head())
```

```python
# =========================
# 최종 검증
# =========================

verify = pd.DataFrame({
    "product_final_품목명": product_final[PRODUCT_NAME_COL],
    "image_final_품목명": image_final[IMAGE_NAME_COL],
    "product_final_제조사명": product_final[PRODUCT_COMPANY_COL],
    "image_final_제조사명": image_final[IMAGE_COMPANY_COL],
})

verify["품목명일치"] = verify["product_final_품목명"] == verify["image_final_품목명"]
verify["제조사명일치"] = verify["product_final_제조사명"] == verify["image_final_제조사명"]

print("행 수 일치:", len(product_final) == len(image_final))
print("품목명 전체 일치:", verify["품목명일치"].all())
print("제조사명 전체 일치:", verify["제조사명일치"].all())

print("품목명 불일치 수:", (~verify["품목명일치"]).sum())
print("제조사명 불일치 수:", (~verify["제조사명일치"]).sum())

print("\n[이미지 매칭 실패 행]")
unmatched = aligned[aligned["이미지매칭여부"] == "N"].copy()

display(
    unmatched[
        [PRODUCT_NAME_COL, PRODUCT_COMPANY_COL]
    ]
)

print("\n[품목명/제조사명 불일치 행]")
display(
    verify[
        (~verify["품목명일치"]) | (~verify["제조사명일치"])
    ]
)
```

```python
# =========================
# 최종에 사용되지 않은 image 행 확인
# =========================

used_image_ids = set(
    aligned["_image_original_row_id"]
    .dropna()
    .astype(int)
    .tolist()
)

image_work["_used_in_final"] = image_work["_image_original_row_id"].isin(used_image_ids)

removed_image_rows = image_work[
    ~image_work["_used_in_final"]
].copy()

print("product_images 원본 행 수:", len(image_df))
print("image_final 행 수:", len(image_final))
print("최종에 사용되지 않은 image 행 수:", len(removed_image_rows))

display(
    removed_image_rows[
        [
            "_image_original_row_id",
            "_before_품목명",
            "_processed_품목명",
            "제조사명",
            "_delete_flag",
            "_preprocess_reason"
        ]
    ].head(200)
)
```

```python
# =========================
# 저장
# =========================

final_image_path = OUTPUT_DIR / "product_images_final_aligned.csv"
preprocessed_image_path = OUTPUT_DIR / "product_images_preprocessed_before_align.csv"
removed_image_path = OUTPUT_DIR / "removed_product_images_rows.csv"
aligned_debug_path = OUTPUT_DIR / "aligned_debug_table.csv"

image_final.to_csv(
    final_image_path,
    index=False,
    encoding="utf-8-sig"
)

image_preprocessed.to_csv(
    preprocessed_image_path,
    index=False,
    encoding="utf-8-sig"
)

removed_image_rows.to_csv(
    removed_image_path,
    index=False,
    encoding="utf-8-sig"
)

aligned.to_csv(
    aligned_debug_path,
    index=False,
    encoding="utf-8-sig"
)

print("저장 완료")
print("최종 image 파일:", final_image_path)
print("전처리 후 align 전 image 파일:", preprocessed_image_path)
print("최종에서 제거된 image 행:", removed_image_path)
print("디버그용 매칭 테이블:", aligned_debug_path)

print("\n최종 확인")
print("product_final 행 수:", len(product_final))
print("image_final 행 수:", len(image_final))
print("행 수 일치:", len(product_final) == len(image_final))
```
